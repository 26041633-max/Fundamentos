```python

# Version 3 - Programa de Prestamos
# -------------------------------------------------------------------------------
# Lista de alumnos (Matriz: [Numero de Control, Nombre, Carrera])
alumnos = [
    ["26040200", "Superman", "Ingeniería en Inteligencia Artificial"],
    ["26040201", "Bat Man", "Ingeniería en Inteligencia Artificial"],
    ["26040202", "Linterna Verde", "Ingeniería Informática"],
    ["26040203", "Mujer Maravilla", "Ingeniería Informática"],
    ["26040204", "Flash", "Ingeniería en Inteligencia Artificial"],
]

# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras = [
    "PC-LC1-01",
    "PC-LC1-02",
    "PC-LC1-03",
    "PC-LC2-01",
    "PC-LC2-02",
    "PC-LC2-03",
    "PC-LC3-01",
    "PC-LC3-02",
    "PC-LC3-03",
    "PC-LC3-04",
]
aulas = ["LC1", "LC1", "LC1", "LC2", "LC2", "LC2", "LC3", "LC3", "LC3", "LC3"]
software = [
    "Python, VS Code",
    "Python, Java, VS Code",
    "AutoCAD, MATLAB",
    "Python, VS Code",
    "Python, Java, VS Code",
    "AutoCAD, MATLAB",
    "Python, VS Code",
    "Python, Java, VS Code",
    "AutoCAD, MATLAB",
    "Python, VS Code",
]
estados = [
    "Disponible",
    "Prestada",
    "Reservada",
    "Disponible",
    "Prestada",
    "Disponible",
    "Disponible",
    "Prestada",
    "Reservada",
    "Mantenimiento",
]

# Listas de control para Historiales
historial_prestamos = []  # Estructura: [num_control, programa, aula, id_comp, carrera]
historial_necesidades = []  # Estructura: [programa, fecha_hora]


def consultar_computadoras():
  print("\n--- COMPUTADORAS ---")
  for i in range(len(computadoras)):
    print(
        "Aula:",
        aulas[i],
        "| ID:",
        computadoras[i],
        "| Software:",
        software[i],
        "| Estado:",
        estados[i],
    )


# MENÚ PRINCIPAL
# Usamos un ciclo while que se repetira indefinidamente hasta que el usuario teclee la opcion 10 (salir)
opcion = 0
while opcion != 10:
  print("\n-----------------------------------")
  print(" SISTEMA DE PRESTAMOS ")
  print("------------------------------------")
  print("1. Consultar computadoras")
  print("2. Buscar por software")
  print("3. Solicitar préstamo")
  print("4. Devolver computadora")
  print("5. Crear reservación")
  print("6. Agregar computadora")
  print("7. Historial de préstamos")
  print("8. Necesidades de Software")
  print("9. Estadísticas")
  print("10. Salir")
  print("--------------------------------------")

  try:
    opcion = int(input("Seleccione una opción: "))
  except ValueError:
    print("\nOpción no válida. Ingrese un número.")
    continue

  # 1. Consultar computadoras
  if opcion == 1:
    consultar_computadoras()

  # 2. Buscar por software
  elif opcion == 2:
    programa = input("\nIngrese el software que desea buscar: ")
    print("\n--- RESULTADOS ---")
    encontrado = False
    for i in range(len(computadoras)):
      if programa.lower() in software[i].lower():
        print("ID:", computadoras[i])
        print("Aula:", aulas[i])
        print("Software:", software[i])
        print("Estado:", estados[i])
        print("-" * 20)
        encontrado = True

    if not encontrado:
      print("No se encontraron computadoras con ese software.")

  # 3. Solicitar préstamo
  elif opcion == 3:
    print("\n--- SOLICITAR PRÉSTAMO ---")
    num_control = input("Numero de control del alumno: ")

    # Búsqueda en la matriz alumnos (recorriendo filas y comparando la primera columna alumnos[i][0])
    alumno_encontrado = None
    for i in range(len(alumnos)):
      if alumnos[i][0] == num_control:
        alumno_encontrado = alumnos[i]
        break

    if alumno_encontrado is not None:
      print(f"Si existe el alumno: {alumno_encontrado[1]}")
      programa = input("Que Programa necesita: ")

      # Verificar si el software existe en alguna computadora
      existe_software = False
      for i in range(len(computadoras)):
        if programa.lower() in software[i].lower():
          existe_software = True
          break

      # Si el programa NO existe en la lista general de software
      if not existe_software:
        confirmar = input(
            f"El programa '{programa}' no existe en el laboratorio. ¿Desea"
            " agregarlo a la lista de necesidades? (s/n): "
        )
        if confirmar.lower() in ["s", "si", "sí"]:
          fecha_actual = datetime.now().strftime("%d/%m/%Y %H:%M:%S")
          historial_necesidades.append([programa, fecha_actual])
          print("Programa agregado a la lista de necesidades.")
        else:
          print("No se agregó a la lista de necesidades.")

      else:
        # El programa SI existe: Intentar asignar una computadora disponible
        asignado = False
        for i in range(len(computadoras)):
          if programa.lower() in software[i].lower():
            if estados[i] == "Disponible":
              estados[i] = "Prestada"
              print("Prestamo registrado correctamente")
              print("Computadora Asignada:", computadoras[i])
              print("Aula Asignada:", aulas[i])

              # Registrar información en el historial de préstamos
              # Guarda: Numero de control, Software, Aula, ID asignado, Carrera
              historial_prestamos.append([
                  num_control,
                  programa,
                  aulas[i],
                  computadoras[i],
                  alumno_encontrado[2],
              ])

              asignado = True
              break

        if not asignado:
          print("No hay computadoras disponibles con ese programa.")

    else:
      print("El alumno no existe en el sistema")

  # 4. Devolver computadora
  elif opcion == 4:
    print("\n--- DEVOLVER COMPUTADORA ---")
    id_comp = input("Ingrese el ID de la computadora a devolver: ")
    if id_comp in computadoras:
      idx = computadoras.index(id_comp)
      if estados[idx] == "Prestada":
        estados[idx] = "Disponible"
        print("Computadora devuelta exitosamente.")
      else:
        print("La computadora no se encuentra en estado 'Prestada'.")
    else:
      print("El ID de la computadora no existe.")

  # 5. Crear reservación
  elif opcion == 5:
    print("\n--- CREAR RESERVACIÓN ---")
    num_control = input("Numero de control del alumno: ")

    # Validar alumno en matriz
    alumno_existe = False
    for i in range(len(alumnos)):
      if alumnos[i][0] == num_control:
        alumno_existe = True
        break

    if alumno_existe:
      id_comp = input("Ingrese el ID de la computadora a reservar: ")
      if id_comp in computadoras:
        idx = computadoras.index(id_comp)
        if estados[idx] == "Disponible":
          estados[idx] = "Reservada"
          print("Reservación creada con éxito.")
        else:
          print("La computadora no está disponible para reservar.")
      else:
        print("El ID de la computadora no existe.")
    else:
      print("El alumno no existe en el sistema")

  # 6. Agregar computadora
  elif opcion == 6:
    print("\n--- AGREGAR COMPUTADORA ---")
    id_computadora = input("ID de la computadora: ")
    aula = input("Aula: ")
    programas = input("Software instalado: ")
    computadoras.append(id_computadora)
    aulas.append(aula)
    software.append(programas)
    estados.append("Disponible")
    print("\nComputadora agregada correctamente.")

  # 7. Historial de préstamos
  elif opcion == 7:
    print("\n--- HISTORIAL DE PRESTAMOS DE COMPUTADORAS ---")
    if len(historial_prestamos) == 0:
      print("No hay préstamos registrados.")
    else:
      for i in range(len(historial_prestamos)):
        p = historial_prestamos[i]
        print(
            f"{i+1} | Alumno: {p[0]} | Software: {p[1]} | Aula: {p[2]} | ID:"
            f" {p[3]}"
        )

  # 8. Necesidades de Software
  elif opcion == 8:
    print("\n--- HISTORIAL DE NECESIDADES ---")
    if len(historial_necesidades) == 0:
      print("No hay necesidades de software registradas.")
    else:
      for i in range(len(historial_necesidades)):
        n = historial_necesidades[i]
        print(f"{i+1} | Programa: {n[0]} | Fecha: {n[1]} |")

  # 9. Estadísticas
  elif opcion == 9:
    print("\n======= Estadísticas por programa ===========")
    if len(historial_prestamos) == 0:
      print("No hay datos registrados en préstamos.")
    else:
      # Conteo por programa
      programas_contados = {}
      carreras_contadas = {}

      for p in historial_prestamos:
        prog = p[1].lower()
        carr = p[4]

        programas_contados[prog] = programas_contados.get(prog, 0) + 1
        carreras_contadas[carr] = carreras_contadas.get(carr, 0) + 1

      for prog, cant in programas_contados.items():
        print(f"{prog} : {cant} veces")

      print("\n======= Estadísticas por carrera ==========")
      for carr, cant in carreras_contadas.items():
        print(f"{carr}: {cant} veces")

  # 10. Salir
  elif opcion == 10:
    print("\nGracias por utilizar el sistema.")
    print("(/\\_/\\)")
    print("( o.o )")
    print("(> ^ <)")
    break
  else:
    print("\nOpción no válida.")
```
